# 02 Summary of the tracked cells per condition

Median diameter, velocity and deformation of the tracked cells per cell line and total flow rate. Each retained track
contributes its median, and bright field (BF) and Ph2 are pooled. The medians of each imaging mode alone are kept as
extra columns. 600 µL/min is excluded (chip failure).

**Input:** `cells_primary.csv` of the cell-tracking run (one row per retained track).
Set the folder below or in the environment variable `CELL_TABLES`.

**Output:** `results/measured/measured_summary.csv`, read by `03_comsol_analysis.ipynb`:

* the median diameter at 300 µL/min selects the simulated particle diameter (15 µm for both cell lines),
* the median diameter of every condition enters the shear stress at the edge of a centred cell (Results,
  Section `sec:res-particles`),
* the median velocity gives the passage time from the channel entrance to the assumed imaging position (Discussion).

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

import measured_summary as ms

pd.set_option('display.precision', 3)
pd.set_option('display.width', 200)

TABLE_DIR = ms.TABLE_DIR          # or Path('<tracking run>/tables')
print('Tables from:', Path(TABLE_DIR).resolve())
```

## 1 Tracks per condition

```python
cells = ms.load_tracks(TABLE_DIR)
display(cells.groupby(['cell_type', 'flow_rate_ul_min', 'imaging_mode']).size().unstack('imaging_mode'))
```

## 2 Pooled summary

`<metric>_median`, `_q25`, `_q75`: median and interquartile range of the track medians, BF and Ph2 pooled.
`<metric>_median_BF` and `_median_Ph2`: median of each imaging mode alone.

```python
summary = ms.run(TABLE_DIR, save=True)
cols = ['cell_type', 'flow_rate_ul_min', 'n_tracks', 'n_BF', 'n_Ph2',
        'equivalent_diameter_um_median', 'equivalent_diameter_um_median_BF', 'equivalent_diameter_um_median_Ph2',
        'speed_m_s_median', 'speed_m_s_median_BF', 'speed_m_s_median_Ph2']
display(summary[cols])
print('written to', ms.SUMMARY_CSV.resolve())
```
