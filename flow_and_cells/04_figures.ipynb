# 04 Figures: chip, flow and cells

All figures of the flow and cell-dynamics part of the thesis, drawn from saved tables with `flow_figures.py`.
Nothing is segmented or tracked here. The figure names are the file names used in the thesis (`images/<name>.png`).

* **Thesis:** `results/figures/thesis/<name>.pdf` and `.png` (300 dpi), with the caption as `<name>_caption.txt`.
  Half-width figures are 7.5 cm wide, full-width figures 15.5 cm, so that the font size in the PDF equals the text size.
* **Talk:** `results/figures/talk/<name>.png`, sans serif with larger fonts.
* **Tables:** `results/figures/tables/*.csv` (`stress_conditions`, `lateral_position_pooled`, `rotation_summary`).

**Inputs**

| Input | Used by | Setting |
|---|---|---|
| `flow_theory.py` | chip and flow figures | none |
| COMSOL exports in `comsol/` | COMSOL figures, `res_flow_velocity*`, `res_pressure_drop`, `res_cellmodel_stress` | environment variable `COMSOL_DIR` |
| `results/measured/measured_summary.csv` | `res_cellmodel_stress` | run `02_measured_summary.ipynb` first |
| `cells_primary.csv`, `detection_morphology.csv` of the tracking run | deformation and rotation figures | `F.CFG.v4_tables` |
| `lateral_position_tracks.csv` | `lateral_position_pooled` | `F.CFG.lateral_csv` |
| static deformation reference (median and IQR of $D_\mathrm{ellipse}$ per cell line) | `res_cif_deformation` | `F.CFG.static_csv` |

If an input is missing, the figure is skipped and the missing file is printed, so the notebook always runs from top
to bottom.
